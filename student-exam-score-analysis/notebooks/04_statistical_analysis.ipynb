# Import Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr

sns.set(style="whitegrid")
```

```python
#File dataset
file_id = "1W0yAusBO4-Hzs2RfQklLW9VSG46T6sxR"
url = f"https://drive.google.com/uc?id={file_id}"

#Baca CSV dari Drive
student_exam_scores = pd.read_csv(url)
```

```python
# Menampilkan 5 baris pertama
student_exam_scores.head()
```

```python
# Statistik deskriptif untuk melihat ringkasan data
print("Statistik Deskriptif:")
print(student_exam_scores[['hours_studied', 'attendance_percent', 'exam_score']].describe())

# Korelasi Pearson
corr_matrix = student_exam_scores[['hours_studied', 'attendance_percent', 'exam_score']].corr(method='pearson')
print("\nMatriks Korelasi:")
print(corr_matrix)

# Visualisasi korelasi dengan heatmap
plt.figure(figsize=(6,4))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Heatmap Korelasi antar Variabel")
plt.show()

# Korelasi detail dengan pearsonr
corr_study, p_study = pearsonr(student_exam_scores['hours_studied'], student_exam_scores['exam_score'])
corr_attend, p_attend = pearsonr(student_exam_scores['attendance_percent'], student_exam_scores['exam_score'])

print(f"\nKorelasi antara Jam Belajar dan Nilai Ujian: {corr_study:.2f}, p-value = {p_study:.4f}")
print(f"Korelasi antara Kehadiran dan Nilai Ujian: {corr_attend:.2f}, p-value = {p_attend:.4f}")
```
