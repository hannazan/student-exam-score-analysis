# Import Data

```python
import pandas as pd
```

```python
#File dataset
file_id = "1hVsQ7GLg2eZoiRZtisDbuAP9qL50h_2Q"
url = f"https://drive.google.com/uc?id={file_id}"

#Baca CSV dari Drive
student_exam_scores = pd.read_csv(url)
```

# Pemeriksaan Data

# Pembersihan Data

```python
#Menghapus kolom yang tidak diperlukan
student_exam_scores = student_exam_scores.dropna(axis=1, how='all')
```

```python
#Melihat apakah terdapat missing values  atau data kosong (null)

student_exam_scores.isnull().sum()
```

```python
#Melihat apakah terdapat data duplikat

student_exam_scores.duplicated().sum()
```

```python
student_exam_scores.describe()
```

```python
num_cols = student_exam_scores.select_dtypes(include=["float64","int64"]).columns

for col in num_cols:
    Q1 = student_exam_scores[col].quantile(0.25)
    Q3 = student_exam_scores[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = student_exam_scores[(student_exam_scores[col] < lower_bound) | (student_exam_scores[col] > upper_bound)]

    print(f"Kolom: {col}, Jumlah Outlier: {len(outliers)}")
```

# Kesimpulan

Dalam proses pengolahan dataset student_exam_scores, beberapa langkah data cleaning dan pengecekan kualitas data telah dilakukan agar data lebih konsisten, representatif, dan siap dipakai untuk analisis lebih lanjut.

**1. Pengecekan ringkasan statistik dengan** `.describe()`

Ringkasan deskriptif digunakan untuk memahami sebaran data pada setiap variabel (hours_studied, sleep_hours, attendance_percent, previous_scores, dan exam_score). Dari sini dapat diketahui nilai minimum, maksimum, kuartil, serta rata-ratanya.

**2. Pengecekan distribusi dan outlier menggunakan metode IQR**

Outlier dicek dengan menghitung batas bawah (Q1 – 1.5 × IQR) dan batas atas (Q3 + 1.5 × IQR). Hasil awal menunjukkan bahwa pada variabel exam_score terdapat kemungkinan nilai ekstrem (misalnya nilai 17.1 yang jauh di bawah Q1). Variabel lain masih berada dalam rentang yang wajar.

**3. Identifikasi fungsi outlier dalam analisis**

Outlier penting untuk diperhatikan karena bisa menunjukkan adanya kesalahan input data, kondisi khusus yang berbeda dari populasi umum, atau informasi penting yang perlu dianalisis terpisah. Dalam konteks penelitian, outlier bisa mempengaruhi model statistik maupun machine learning jika tidak ditangani dengan tepat.

**4. Penanganan awal outlier**

Strategi umum yang dapat digunakan meliputi:

*   Menghapus outlier jika terbukti sebagai kesalahan pencatatan.
*   Mengganti (imputasi) dengan median agar distribusi data tetap stabil.
*   Membiarkan outlier jika memang mencerminkan kondisi riil (misalnya siswa dengan nilai ujian sangat rendah karena faktor khusus).

**5. Ekspor data hasil cleaning ke CSV/Excel.**

Dataset hasil olahan disimpan kembali agar mudah dibagikan dan digunakan ulang.
