# Import Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
#File dataset
file_id = "1W0yAusBO4-Hzs2RfQklLW9VSG46T6sxR"
url = f"https://drive.google.com/uc?id={file_id}"

#Baca CSV dari Drive
student_exam_scores = pd.read_csv(url)
```

# Insight 1: Jam belajar vs Nilai Ujian
## "Semakin banyak jam belajar, semakin tinggi nilai ujian siswa"

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x="hours_studied", y="exam_score", data=student_exam_scores, color="blue", alpha=0.7)
sns.regplot(x="hours_studied", y="exam_score", data=student_exam_scores, scatter=False, color="red")

plt.title("Semakin banyak jam belajar, semakin tinggi nilai ujian siswa", fontsize=14)
plt.xlabel("Jam Belajar (hours_studied)")
plt.ylabel("Nilai Ujian (exam_score)")
plt.show()

print("Insight 1: Jika saya tunjukkan ini ke stakeholder, keputusan yang bisa dia ambil adalah pentingnya meningkatkan jam belajar siswa agar nilai ujian lebih baik.")
```

# Insight 2: Kehadiran vs Nilai Ujian
## "Kehadiran di kelas berhubungan dengan nilai ujian, siswa dengan >80% kehadiran cenderung memiliki nilai lebih baik"


```python
plt.figure(figsize=(8,6))
sns.scatterplot(x="attendance_percent", y="exam_score", data=student_exam_scores, color="green", alpha=0.7)
sns.regplot(x="attendance_percent", y="exam_score", data=student_exam_scores, scatter=False, color="red")

plt.axvline(80, color="orange", linestyle="--", label="Batas 80% Kehadiran")
plt.legend()

plt.title("Kehadiran >80% berhubungan dengan nilai ujian lebih baik", fontsize=14)
plt.xlabel("Persentase Kehadiran (%)")
plt.ylabel("Nilai Ujian (exam_score)")
plt.show()

print("Insight 2: Jika saya tunjukkan ini ke stakeholder, keputusan yang bisa dia ambil adalah meningkatkan kehadiran siswa di kelas agar nilai ujian meningkat.")
```

# Insight 3: Nilai Ujian Sebelumnya vs Nilai Ujian Sekarang
## "Nilai ujian sebelumnya cukup berpengaruh terhadap nilai ujian terbaru"

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x="previous_scores", y="exam_score", data=student_exam_scores, color="purple", alpha=0.7)
sns.regplot(x="previous_scores", y="exam_score", data=student_exam_scores, scatter=False, color="red")

plt.title("Nilai ujian sebelumnya memengaruhi nilai ujian terbaru", fontsize=14)
plt.xlabel("Nilai Ujian Sebelumnya")
plt.ylabel("Nilai Ujian Sekarang (exam_score)")
plt.show()

print("Insight 3: Jika saya tunjukkan ini ke stakeholder, keputusan yang bisa dia ambil adalah memperhatikan nilai ujian sebelumnya sebagai indikator awal performa siswa.")
```
